# First programs

We will start with some basic input / output operations in this. As with the first programming lesson in any other programming language, we will start with a "Hello World!" program and go to input function afterwards.

## Basic input and output functions

input -> The program asks for some data from the user, and the user inputs that to the program

output -> The program performs some operations and outputs the results to the user 

### Cell - 1

```python
# My first python program
# This will print Hello World on the screen
# print is a function that outputs text to the screen
# You are calling the print function and passing it a string of characters to print on the screen
print ('Hello World!')

# In this program, the first three lines are comments
# Anything starting with '#' is a comment and is ignored by the Python interpreter
# However, comments are useful for humans to understand the code

# The fourth line is a print statement to print something on the screen.
# Here, we are printing the words "Hello" and "World" followed by "!"
# Note that these words are inside quotes - to indicate that it is a STRING of characters.
```

### Cell - 2

```python
# In these programs, "print" is an instruction, which takes some arguments in a bracket
# You could try more on your own
# e.g. print your name.
print ('My name is John Doe')

# e.g. print what you are learning
print ('I am learning Python programming!')

# e.g. print numbers
print (2026)

# Try changing the text within the quotes and run the cell again!
```

### Cell - 3

```python
# Input operations
# You can also take input from the user using the input() function
name = input("Enter your name: ")  # This will prompt the user to enter their name

# The data entered by the user is stored in the variable 'name'
print("Hello,", name)  # This will print a greeting to the user with their name

# Try running the cell and entering your name when prompted!
# Look at the top of the VS Code window, you will see a small terminal window
# where you can enter your name.
```
