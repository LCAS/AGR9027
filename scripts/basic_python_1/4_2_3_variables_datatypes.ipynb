# Variables and Datatypes

## Variables
Variables are an important concept in any programming langugage. They are generally used to represent a value stores in the memory, to be accessed, modified and reused.

In Python, a variable name can be a string, starting with a character or some symbols and not a number, and should not contain any spaces in between.

You can directly initiate a variable to a particular data type or by assigning a value to it.

## Important note!!!
Do not create a variable with the same name as special functions in Python. The functions will not be accessible afterwards while running the program.


```python
# A variable is like a box where you can store things, like numbers or words.
# You can give each box a name and then put information inside.

# Storing numbers
height = 5.5  # 'height' is the name of the box and it holds the number 5.5
print('My height is ', height, ' feet.')

# Storing words
name = "Tom"
print ('My name is ', name)

#Try changing the variable names and the values stored in them!
# for example, you can use different variable names and store different values in it.
# Then print them out!
my_name = "Alice"
my_height = 6.0
print('My name is ', my_name, ' and my height is ', my_height, ' feet.')


# Try changing the variable names and the values stored in them!
```

## Data types in Python
Data types indicate the nature of the value you can store in a variable. The data types in Python can be split into two groups - the basic ones and the complex ones. 

The basic data types are:

1. Character
2. String
3. Boolean
4. Integer
5. Float
6. Complex number
7. None

Some other data types are also available to store a group of values. They are:
1. Tuple
2. List
3. Dictionary
4. Set

### Character data type

```python
# Character data type
# The character data type is used to store single characters like letters, digits, or symbols.

# A single character
char_variable = 'A' 
print(char_variable)

# unicode characters
# unicode characters can be used to store characters from different languages or even special symbols.
# smiley face
unicode_variable = '😊'
print(unicode_variable)

# spanish character
unicode_variable_spanish = 'ñ'
print(unicode_variable_spanish)

# Try changing the variable names and the values stored in them!
```

### String data type

```python
# Strings - A string is a sequence of characters enclosed in single or double quotes.

# You can use either single quotes (' ') or double quotes (" ") to define a string in Python.
# However, make sure to use the same type of quotes at the beginning and end of the string.

# A sequence of characters
string_variable = "Hello" 
print(string_variable)

# You can also use single quotes
another_string_variable = 'World' 
print(another_string_variable)

# unicode strings
unicode_string_variable = "Café" 
print(unicode_string_variable)

# A number stored as a string
number_string = '2024' 
print(number_string)
# However, you cannot perform mathematical operations on number_string unless you convert it to a numeric type.

# Try changing the variable names and the values stored in them!
```

### Boolean data type

```python
# Boolean data type -> True or False
is_sunny = True  # This variable holds a boolean value
print(is_sunny)

is_raining = False  # This variable holds another boolean value
print(is_raining)

# Try changing the variable names and the values stored in them!
```

### Numeric datatypes 
* Integer
* Float
* Complex

```python
# Numeric data type (1) -> Integer

# Integer data type
# An integer is a whole number without a decimal point
age = 25
print ('My age is ', age, ' years.')
# An integer can be positive or negative
temperature = -5
print ('The temperature is ', temperature, ' degrees Celsius.')

# Alternatively, you can also create integers using the int() function
age = int(25)
print ('My age is ', age, ' years.')

# We used number strings before, you can convert them to integers using int()
number_string = '2025'
year = int(number_string)
print('The year is ', year)

# You can perform mathematical operations with integers
sum_result = age + 5
print('In 5 years, I will be ', sum_result, ' years old.')

# Try changing the variable names and the values stored in them!
```

```python
# Numeric data type (2) -> Float

# Float data type
price = 19.99
print ('The price is ', price, ' pounds.')

# A float can also be created using the float() function
weight = float(70.5)
print ('The weight is ', weight, ' kg.')

# You can also use scientific notation to represent very large or very small numbers
distance = 1.5e6  # This represents 1.5 million
print ('The distance is ', distance, ' meters.')

# You can use float() function to create infinity and NaN (Not a Number)
infinity_value = float('inf')
nan_value = float('nan')
print('Infinity value: ', infinity_value)
print('NaN value: ', nan_value)

# You can perform mathematical operations with floats
total_cost = price + 5.50
print('The total cost is ', total_cost, ' pounds.')

# Try changing the variable names and the values stored in them!
```

```python
# Numeric data type (3) -> Complex

# Complex data type
complex_number_1 = 2 - 3j
print('The first complex number is ', complex_number_1)

# A complex number has a real part and an imaginary part. In this case, 2 is the real part and -3 is the imaginary part.
# You can access the real and imaginary parts using the .real and .imag attributes
print('Real part: ', complex_number_1.real)
print('Imaginary part: ', complex_number_1.imag)

# Alternatively, you can also create complex numbers using the complex() function
complex_number_2 = complex(2, 8)
print('The second complex number is ', complex_number_2)

# You can do arithmetic operations with complex numbers
complex_sum = complex_number_1 + complex_number_2
print('Sum of complex numbers: ', complex_sum)
```

```python
# Converting between numeric data types

# Converting an integer to a float
integer_number = 10
float_number = float(integer_number)
print('The float value of ', integer_number, ' is ', float_number)

# Converting a float to an integer (this will truncate the decimal part)
float_number = 9.99
integer_number = int(float_number)
print('The integer part of ', float_number, ' is ', integer_number)

# Converting an integer to a complex number
integer_number = 7
complex_number = complex(integer_number)
print('The complex value of ', integer_number, ' is ', complex_number)

# Converting a float to a complex number
float_number = 4.5
complex_number = complex(float_number)
print('The complex value of ', float_number, ' is ', complex_number)

# converting a complex number to a float or integer is not directly possible and will raise an error.
# You can only extract the real or imaginary part using .real or .imag attributes.

# Converting a number string to an integer
number_string = '123'
integer_number = int(number_string)
print('The integer value of the string "', number_string, '" is ', integer_number)

# Converting a number to a string
number = 456
number_string = str(number)
print('The string value of the number ', number, ' is "', number_string, '"')

# Try changing the variable names and the values stored in them!
```

```python
# You can use math module functions on these numbers as well.
import math

# You can round a float to the nearest integer
rounded_number = round(9.56)
print('The rounded value of 9.56 is ', rounded_number)
# or use math.floor and math.ceil
floored_number = math.floor(9.56)
ceiled_number = math.ceil(9.56)
print('The floored value of 9.56 is ', floored_number)
print('The ceiled value of 9.56 is ', ceiled_number)


# Working with trigonometric functions
angle = 45 # degrees
# let's get sine and cosine of this angle
# the sin and cos function in math takes radians, not degrees
angle_radians = math.radians(angle)
sine_angle = math.sin(angle_radians)
cosine_angle = math.cos(angle_radians)
print ('Sin (%0.3f) is %0.3f' %(angle, sine_angle))
print ('Cosine (%0.3f) is %0.3f' %(angle, cosine_angle))
```

```python
# You can use the function type() to check the data type of a variable
price = 19
weight = 70.5
a_complex = 3 + 4j
print('The data type of price is ', type(price))
print('The data type of weight is ', type(weight))
print('The data type of a_complex is ', type(a_complex))
```
