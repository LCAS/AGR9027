# Lists

We already saw some details about lists in the previous notebook. We are just going through some more details here.

The scripts here demonstrate various operations that can be performed on a list in Python.

```python
# Let's say we have a list here
a_list = [1.2, 2.0, 4.2, -1.4, 5.9, 1.2, -1.4]

# you can get the number of items in a list using len()
length = len(a_list)
print("Length of the list:", length)

# You can access elements by their index
# remember that indexing starts at 0 in Python
first_element = a_list[0]
print("First element:", first_element)

# You can also access the last element using negative indexing
last_element = a_list[-1]
print("Last element:", last_element)

# you can get the sum of all elements in the list using sum()
total_sum = sum(a_list)
print("Sum of all elements:", total_sum)

# You can find the maximum and minimum values in the list using max() and min()
max_value = max(a_list)
min_value = min(a_list)
print("Maximum value:", max_value)
print("Minimum value:", min_value)

# you can sort the list using the sorted() function
sorted_list = sorted(a_list)
print("Sorted list:", sorted_list)

# You can also reverse the list using the reversed() function
reversed_list = list(reversed(a_list))
print("Reversed list:", reversed_list)

# you can slice the list to get a sublist of elements
sublist = a_list[1:4]  # gets elements from index 1 to 3
print("Sublist of the list (index 1 to 3):", sublist)

print (2 * sublist)
```

## finding statistics of numbers in a list

```python
# Although lists do not have built-in statistical methods, you can use the statistics module for more advanced operations
# import the statistics module
import statistics

# Let's say we have a list here
a_list = [1.2, 2.0, 4.2, -1.4, 5.9, 1.2, -1.4]

mean_value = statistics.mean(a_list)
median_value = statistics.median(a_list)
print("Mean value:", mean_value)
print("Median value:", median_value)
```

## converting a list to a tuple and a set

```python
# Let's say we have a list here
a_list = [1.2, 2.0, 4.2, -1.4, 5.9, 1.2, -1.4]

# You can convert a list to a tuple using the tuple() function
a_tuple = tuple(a_list)
print("Tuple converted from list:", a_tuple)

# You can convert a list to a set using the set() function
a_set = set(a_list)
print("Set converted from list:", a_set)
```

## converting a list to a dictionary

```python
# you can convert a list to a dictionary if it contains pairs of elements
# For example, let's create a list of pairs
pairs_list = [("a", 1), ("b", 2), ("c", 3)]
a_dict = dict(pairs_list)
print("Dictionary converted from list of pairs:", a_dict)
```

## concatenate two lists

```python
# You can concatenate two lists using the + operator
first_list = [10, 20, 30]
second_list = [40, 50, 60]
concatenated_list = first_list + second_list
print("Concatenated list:", concatenated_list)

# you can repeat a list multiple times using the * operator
repeated_list = first_list * 3
print("Repeated list:", repeated_list)
# This is similar to concatenating the list to itself multiple times
```
