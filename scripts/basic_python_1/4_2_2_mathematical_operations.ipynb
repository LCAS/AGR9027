# Mathematical Operations in Python

You can use python to perform different mathematical operations.

* Addition (**+**)
* Subtraction (**-**)
* Multiplication (**\***)
* Division (**/**)
* Remainder (**%**)
* Power (**\*\***)

Some examples are shown below. You can modify them and try them out.

### Cell - 1

```python
# Mathematical operations in Python - 1
# In here, we will do some maths and also leave some in-line comments

# Addition (+), Subtraction (-), Multiplication (*), Division (/), Remainder (%) and Power (**)

# Adding numbers
print(5 + 3)  # This will show 8

# Subtracting numbers
print(10 - 4)  # This will show 6

# Multiplying numbers
print(2 * 3)  # This will show 6

# Dividing numbers
print(9 / 3)  # This will show 3.0

# Remainder (modulus)
print(7 % 3)  # This will show 1

# Power (exponentiation)
print(2 ** 3)  # This will show 8

# You can also use decimals (floating point numbers)
print(5.5 + 2.2)  # This will show 7.7

# Try changing the numbers and operations to see different results!

# You can improve the output by adding text to explain the result
print("The result of 5 + 3 is:", 5 + 3)
# Here, we are printing a string followed by the result of the addition

# Try changing the text and the operation to see different outputs!
```

### Cell - 2

```python
# Mathematical operations in Python - 2a
# In here, we will see how to combine multiple operations

# You can also combine operations
print((5 + 3) * 2)  # This will show 16
# Remember the order of operations (BODMAS/BIDMAS)
# BODMAS: Brackets, Orders (i.e. powers and square roots, etc.), Division and Multiplication, Addition and Subtraction
# BIDMAS: Brackets, Indices, Division and Multiplication, Addition and Subtraction
# So, in the above example, (5 + 3) is calculated first, then multiplied by 2

# Try changing the numbers, brackets and operations to see different results!
```

### Cell - 3

```python
# Mathematical operations in Python - 2b
# A more complex example

result = (10 - 2) ** 2 / 4 + 3 * (6 % 4)
print("The final result is:", result)  # This will show 19.0
# Here, we are combining subtraction, power, division, multiplication, and modulus operations to get a final result

# Try changing the expression to see how it affects the final result!
```

### Cell - 4

```python
# Mathematical operations in Python - 3
# Using the math module for advanced operations

# In Python, many advanced functionalities are available through modules
# An example is the math module, which provides access to mathematical functions
# One can import a module to the program using the import statement to access its functions
# Here, we will see some examples of using the math module

# You can import math module for more complex mathematical operations
import math

# Using the sqrt function from the math module to calculate square root
print("The square root of 16 is:", math.sqrt(16))  # This will show 4.0

# Using the factorial function from the math module to calculate factorial
print("The factorial of 5 is:", math.factorial(5))  # This will show 120

# Using the pi constant from the math module
print("The value of pi is approximately:", math.pi)  # This will show 3.141592653589793

# Try using other functions from the math module by exploring its documentation!
# You can find the documentation here: https://docs.python.org/3/library/math.html
```
