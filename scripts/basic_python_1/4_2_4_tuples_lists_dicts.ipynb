## Sequence and Mapping data types

There are two data types for storing a sequence of data.
* String -- We have already seen this (a sequence of characters)
* Tuple
* List

A special feature of these sequences is that we can access specific items in that ordered sequence. To access specific item, you have to use the variable name and the position index number. This indexing start from 0 at the left most position, and is incremented by 1 for the next item on right.

You can also use negative indices, with -1 to indicate the right most entry in the sequence, and -2 the second right most entry.

You can also use slicing with the indices, to get parts of the sequences.

There is a data type to store Key:Value pairs. In this, the unique keys map to unique values as defined.
* Dictionary

## Tuple


```python
# Tuple data type

# A tuple is a collection of items which is ordered and immutable (cannot be changed).
# Items are separated by commas and kept inside "(" and ")"
t_fruits = ("apple", "banana", "cherry")
print (t_fruits)

# We can access items in a tuple using indexing (starting from 0)
first_fruit = t_fruits[0]
print("The first fruit is: ", first_fruit)

# We can get the length of a tuple using len() function
print ("length of t_fruits: ", len(t_fruits))

# But tuples are immutable, meaning it cannot be modified in parts, once we define it.
# So we cannot do things like this:
# t_fruits[0] = "orange"
# This will give an error:
# TypeError: 'tuple' object does not support item assignment

# You can use indexing and slicing to access elements in a tuple.
# For example:
print("The first two fruits are: ", t_fruits[0:2])

# you can also use negative indexing to access elements from the end of the tuple.
print("The last fruit is: ", t_fruits[-1])
print("The last two fruits are: ", t_fruits[-2:])


# nested tuples
# -------------


# Tuples can be nested, meaning you can have a tuple inside another tuple.
nested_tuple = (1, 2, (3, 4))
print("The nested tuple is: ", nested_tuple)

# You can access elements in the nested tuple using multiple indices.
print("The innermost element is: ", nested_tuple[2][1])

# You can also use slicing to access elements in the nested tuple.
print("The last element of the nested tuple is: ", nested_tuple[2][-1]) 

# You can delete a tuple entirely using the del keyword.
t_to_delete = (1, 2, 3)
print("Tuple before deletion: ", t_to_delete)
del t_to_delete
# print("Tuple after deletion: ", t_to_delete)  # This will give an error since the tuple is deleted.


# mixed data types in a tuple
# ----------------------

mixed_tuple = (1, "apple", 3.14, True)
print("The mixed tuple is: ", mixed_tuple)


# more tuple operations
# ---------------------

# You can use various built-in functions with tuples.
# For example, you can use the count() method to count how many times an item appears in a tuple.
t_numbers = (1, 2, 3, 2, 4, 2)
count_of_2 = t_numbers.count(2)
print("The count of 2 in t_numbers is: ", count_of_2)

# You can also use the index() method to find the first occurrence of an item in a tuple.
index_of_2 = t_numbers.index(2)
print("The index of the first occurrence of 2 in t_numbers is: ", index_of_2)


# Try modifying the above and see how tuples behave differently
```

## List

```python
# List data type

# A List is a collection of items which is ordered and mutable (can be changed).
# Lists are defined with '[' and ']' at either ends of items separated by commas
fruits = ["apple", "banana", "cherry"]
print (fruits)

# You can get the total number of items in the List
print ("Length of the list fruits: ", len(fruits))

# You can access items using their position index
print ("Item in position 0: ", fruits[0])
print ("Item in position 1: ", fruits[1])

# List items can be modified
print ("Original list: ", fruits)
# Modifying the first item
fruits[0] = "pear"
print ("Modified an item: ", fruits)

# List can be extended by appending a value at the end
print ("Original list before appending: ", fruits)
fruits.append("clementine")
print ("Appended a value: ", fruits)

# List can have multiple copies of the same item
fruits.append("clementine")
print ("Appended an item again: ", fruits)

# We can get the position of a known item in a List
banana_index = fruits.index("banana")
print ("index of banana: ", banana_index)

# We can insert a new item at a specific location in a List
print ("List before insertion: ", fruits)
fruits.insert(2, "pineapple")
print ("Inserted pineapple at index 2: ", fruits)

# you can remove a value form the list. The first instance from the left will be removed, if multiples of that item is present
print ("List before removals: ", fruits)
fruits.remove("clementine")
print ("Removed clementine: ", fruits)
fruits.remove("pear")
print ("Removed pear: ", fruits)

# You can pop out the item at the last position
# where pop() removes the item and also returns it
item = fruits.pop()
print ("popped item from the list: ", item)
print ("list after one pop operation: ", fruits)

# You can also pop out an item at a specific index
item_at_index_1 = fruits.pop(1)
print ("popped item at index 1: ", item_at_index_1)
print ("list after popping item at index 1: ", fruits)

# You can use various built-in functions with Lists.
# For example, you can use the count() method to count how many times an item appears in a List.
l_numbers = [1, 2, 3, 2, 4, 2]
count_of_2 = l_numbers.count(2)
print("The count of 2 in l_numbers is: ", count_of_2)

# You can also use the index() method to find the first occurrence of an item in a List.
index_of_2 = l_numbers.index(2)
print("The index of the first occurrence of 2 in l_numbers is: ", index_of_2)

# You can also use the index() method to find the first occurrence of an item in a List.
index_of_2 = l_numbers.index(2)
print("The index of the first occurrence of 2 in l_numbers is: ", index_of_2)

# you can use various built-in functions such as reverse(), sort(), clear() etc. with Lists.
l_numbers.reverse() # this will reverse the list
print("Reversed l_numbers: ", l_numbers)

l_numbers.sort() # this will sort the list in ascending order
print("Sorted l_numbers: ", l_numbers)

l_numbers.clear() # this will remove all items from the list
print("Cleared l_numbers: ", l_numbers)

# Try modifying the above and see how lists behave differently from tuples
```

## Dictionary

```python
# Dictionary data type
# Mapping type - A value mapped to a key

# Think about some variables to descibe a person
# We can keep all of them together in a Dictionary data type
# Dictionary is defined as a collection of Key:Value pairs separated by "comma", within '{' and '}'
person = {'name': 'John Doe',
          'age': 34,
          'height': 170}
print(person)


# You can access each value, using the variable name and the key name
print (person['name'], person['age'], person['height'])

# You can modify any value in the Dictionary by referring to its key name
person['age'] = 35
print("After modifying age: ", person)
# Alternatively, you can use the update() method to modify multiple key:value pairs at once
person.update({'age': 36, 'height': 172})
print("After updating age and height: ", person)

# You can delete a key:value pair using the del keyword
del person['height']
print("After deleting height: ", person)

# You can add a new key:value pair
person['weight'] = 70
print("After adding weight: ", person)
# You can add another key:value pair
person['city'] = 'London'
print("After adding city: ", person)


person.keys()  # This will give you all the keys in the dictionary
person.values()  # This will give you all the values in the dictionary
person.items()  # This will give you all the key:value pairs in the dictionary

print (person.keys())
print (person.values())
print (person.items())

# You can use pop() method to remove a key:value pair and get the value
age = person.pop('age')
print("Popped age: ", age)
print("Dictionary after popping age: ", person)

# Try changing the variable names and the values stored in them!
```
