In this notebook, we will look at ML for classification with the classical Iris dataset. We mainly use the ML libraries available in _sklearn_module here. Many classical datasets and many ML algorithms are available in the _sklearn_ module. Initially, we will use the Decision Tree algorithm, and then use k-nearest neighbour algorithm for this classification.

Let's look at the Iris dataset first, and see its details

```python
# 1. Load the iris dataset
# iris dataset is available in sklearn (scikit-learn) module

from sklearn import datasets

iris_dataset = datasets.load_iris()
iris_features = iris_dataset.data
feature_names = iris_dataset.feature_names
iris_types = iris_dataset.target
type_names = iris_dataset.target_names

print (iris_features.shape)
print ('feature_names: ', feature_names)
print ('type_names: ', type_names)
```

Let's split the dataset to training and test sets. 
The training set will be used for training the ML models. The ML models have not seen the test set. So, the test set will be used for testing the ML models and the predicted output can be compared against the actual values to evaluate the model performance using metrics like prediction accuracy.
A 70-30 split is done here.

```python
# test - train split
# 70 - 30 split here
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(iris_features, iris_types, test_size=0.3)
```

Let's try Decision Tree first

```python
# classification - 1
# Decision tree
from sklearn import tree

clf = tree.DecisionTreeClassifier() # create a decision tree object
clf = clf.fit(X_train, y_train) # fit the train data

# Plot the tree
tree.plot_tree(clf)

# test the accuracy
acc_dt = clf.score(X_test, y_test)
print ('prediction accuracy using DT = %0.3f' %(acc_dt))
```

Let's try kNN now

```python
# classification - 2
# K Nearest Neighbours
from sklearn import neighbors

n_neighbours = 15
# we create an instance of Neighbours Classifier and fit the data.
clf = neighbors.KNeighborsClassifier(n_neighbours, weights='distance')
clf.fit(X_train, y_train)

# test the accuracy
acc_knn = clf.score(X_test, y_test)
print ('prediction accuracy using KNN = %0.3f' %(acc_knn))
```
