# Assigning values

In the scripts so far, we have assigned values to different variables at numberous occasions.

In this notebook, we will see assigning a value and comparing values

````python
# To assign a value to a variable, we use the "=" sign.
# note that, we use a single "=" sign

# for example, we can assign a string value to a variable as follows
a_str = "this is a string value"

# similarly, we can assign a numeric value to a variable
a_num = 42.6

# we can also assign a boolean value to a variable
a_bool = True
# note that, boolean values can be either True or False
another_bool= False

# We can print the values of these variables to verify the assignments
print("String value:", a_str)
print("Numeric value:", a_num)
print("Boolean value 1:", a_bool)
print("Boolean value 2:", another_bool)
````

````python
# However there are few more assignment operations in Python
# For example, you can use the following
# +=, -=, *=, **=, /=

var_a = 2.0
var_b = 3.0

# let's try those operators here
# +=
var_b += var_a
# this is equivalent to var_b = var_b + var_a
# i.e. the current values of var_a and var_b are added and the result is saved to var_b
# var_b = 2.0 + 3.0
# after this, var_b will have a value of 5.0
print (var_b)

# -=
# can you guess what would happen here?
var_b -= var_a
# var_a will be subtracted from var_b and the result will be stored in var_b
# var_b should have 3.0 after this operation (5.0 - 2.0)
print (var_b)

# *=
var_b *= var_a
# This is equivalent to var_b = var_b * var_a
# var_b will have 6.0 after this operation (3.0 * 2.0)
print (var_b)

# /=
var_b /= var_a
# This is equivalent to var_b = var_b / var_a
# var_b will have 3.0 after this operation (6.0 / 2.0)
print (var_b)

# **=
var_b **= var_a
# This is equivalent to var_b = var_b ** var_a
# var_b will have 9.0 after this operation (3.0 ** 2.0)
print (var_b)
````

# Comparing values

We can compare the value of a variable to another value or value of another variable

This is usually done using comparison operations, which result in a boolean value (`True` or `False`)

Usually we use the following operators for this: `<`, `>`, `<=`, `>=`, `==`, `!=`

`<`  less than  
`>`  greater than  
`<=`  less than or equal to  
`>=`  greater than or equal to  
`==`  equal to  
`!=`  not equal to

## comparing integers

````python
# let's intialise two integers
first_int = 45
second_int = 46

# let's print these values first
print ("first and second intergers: ", first_int, " and ", second_int)

# let's compare these two variables now.
print ("first_int < second_int: ", first_int < second_int)
print ("first_int <= second_int: ", first_int <= second_int)
print ("first_int > second_int: ", first_int > second_int)
print ("first_int >= second_int: ", first_int >= second_int)
print ("first_int == second_int: ", first_int == second_int)
print ("first_int != second_int: ", first_int != second_int)

# Try changing the values to be the same and run the cell
````

## comparing float values

````python
# You have to be careful when comparing float variables and values
# Due to discretisation, a float value may be saved differently to the rounded value that was assigned.

# For example, if you store a number 14.5, it may be stored as 14.4999999999 or 14.5000000001
# However, this may not be an issue with integers
a_float_var = 30.0

# After a series of operations, you might be getting a value such as
another_float_var = 30.00000000001
yet_another_float_var = 29.999999999999

# However, if you want to compare to see if the value is correct, you may get a wrong answer
print ("a_float_var == another_float_var: ", a_float_var == another_float_var)

# instead you can use isclose() function from the math module
import math
print ("Using isclose():")
print ("a_float_var and another_float_var are close: ", math.isclose(a=a_float_var, b=another_float_var, abs_tol=1e-9,))
print ("a_float_var and yet_another_float_var are close: ", math.isclose(a=a_float_var, b=yet_another_float_var, abs_tol=1e-9,))
````

# IF-ELIF-ELSE

Often, we can use the conditions to decide what we need to do next.

Python provides If-elif-else condition checking for this.  
This uses some conditions to decide whether to do a block of code.  
The comparisons we saw above, can be used for these conditions  
These conditions will result in a boolean state (True or False), and  
depending on that the program will decide whether to execute some blocks of code

```
if (condition_1 == True):
    # when condition_1 is True
    block_1 operations 

elif (condition_2 == True): # or use elif (condition_2)
    # when condition_1 is False and condition_2 is True
    block_2 operations

elif (condition_3):
    # when condition_1 is False, condition_2 is False and condition_3 is True
    block_3 operations 

elif (not condition_4):
    # we are explicitly looking for a False state of a condition
    # when condition_1 is False, condition_2 is False, condition_3 is True and condition_4 is False
    block_3 operations 

# you can have more elif condition checks
...

else:
    # For conditions which fail all the above conditions fails
    block_n operations
```

In the if-elif-else structure, elif and else are optional.

Let's see this in practice.


````python
# IF statement
first_int = int(input("Please enter an integer value: "))
second_int = int(input("Please enter another integer value: "))

if first_int < second_int:
    # if the first integer is less than the second integer, print this message
    print ("The first integer is less than the second integer.")

# Did you notice that there is a colon (:) at the end of the if statement?
# This indicates the start of a block of code that will be executed if the condition is true
# The block of code is indented (by 4 spaces or a tab) to indicate that it belongs to the if statement
````

````python
# Now, let's see an example of if-else statement
# Let's get the age of the user and print a message saying whether the user is a minor or an adult.

age = int(input("Please enter your age: "))

if age < 18:
    print ("You are a minor.")

else:
    print ("You are an adult.")
````

````python
# Now, let's see an example of if-elif-else statement
# Let's get the marks of a student and print the grade based on the marks
marks = int(input("Please enter the marks: "))

if marks >= 70:
    print ("You passed the module with a Distinction!!!")

elif marks >= 60:
    print ("You passed the module with a Merit!!!")

elif marks >= 50:
    print ("You passed the module!!!")

else:
    print ("You failed the module. Better luck in the resit!")
````

````python
# However, what would happen if we reorder the conditions?
# Try changing the order of the conditions and see what happens
# For example, let's put the condition for Merit first
marks = int(input("Please enter the marks: "))
if marks >= 60:
    print ("You passed the module with a Merit!!!")

elif marks >= 70:
    print ("You passed the module with a Distinction!!!")

elif marks >= 50:
    print ("You passed the module!!!")

else:
    print ("You failed the module. Better luck in the resit!")

# This program will not work as expected
# because the first condition (marks >= 60) will be true for all marks >= 60
# and the subsequent conditions will never be checked
# An example, if the user enters 75, the first condition will be true and the 
# program will print "You passed the module with a Merit!!!"
# Therefore, it is important to order the conditions correctly
````

# Checking multiple conditions
You can combine different conditions together using boolean operators: AND, OR, NOT, XOR etc

````python
# Let's correct the conditions using multiple conditions with AND operator
# The AND operator is represented by "and" keyword in Python
marks = int(input("Please enter the marks: "))

if (marks >= 50) and (marks < 60):
    print ("You passed the module!!!")

elif (marks >= 60) and (marks < 70):
    print ("You passed the module with a Merit!!!")

elif marks >= 70:
    print ("You passed the module with a Distinction!!!")

else:
    print ("You failed the module. Better luck in the resit!")

# We have seen a better way to do this without using multiple conditions earlier
# by ordering the conditions from highest to lowest
# This way, we do not need to use multiple conditions with AND operator
# and the additional computational cost associated with it
````
