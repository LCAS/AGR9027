In this notebook, we will look at ML for regression with the classical Iris dataset to predic the petal width, given the petal length. We mainly use the ML libraries available in _sklearn_module here. Initially, we will use the Linear Regression algorithm, and then use MLP (multi-layer perceptron or multi-layer neural network) for this regression.

Let's look at the Iris dataset first, and see its details

```python
# 1. Load the iris dataset
# iris dataset is available in sklearn (scikit-learn) module

from sklearn import datasets

iris_dataset = datasets.load_iris()
iris_features = iris_dataset.data
feature_names = iris_dataset.feature_names
iris_types = iris_dataset.target
type_names = iris_dataset.target_names

print (iris_features.shape)
print ('feature_names: ', feature_names)
print ('type_names: ', type_names)
```

Let's split the dataset to training and test sets. 
The training set will be used for training the ML models. The ML models have not seen the test set. So, the test set will be used for testing the ML models and the predicted output can be compared against the actual values to evaluate the model performance using metrics like prediction accuracy.
A 70-30 split is done here.

```python
# train - test split
# 70 - 30 split here (randomly selected 70% for training and the remaining 30% for testing)
from sklearn.model_selection import train_test_split
import numpy

X_train, X_test, y_train, y_test = train_test_split(iris_features, iris_types, test_size=0.3)

petal_length_train = numpy.array(X_train[:,2]).reshape(-1,1) # reshaping because we are considering only one input feature
petal_width_train = numpy.array(X_train[:,3])

petal_length_test = numpy.array(X_test[:,2]).reshape(-1,1)
petal_width_test = numpy.array(X_test[:,3])
```

Before we start the ML modelling, it is good to look at the dataset and see the relationship between the input and output. Here, the relationship between the petal length and petal width is of our interest.

```python
# 2. Visually inspect at correlation between two attributes using scatter plots
# petal length (index=2) vs petal width (index=3)
import matplotlib.pyplot as plt

fig = plt.figure()
ax = fig.add_subplot(111)
ax.scatter(iris_features[0:50,2], iris_features[0:50,3], color="r")
ax.scatter(iris_features[50:100,2], iris_features[50:100,3], color="g")
ax.scatter(iris_features[100:150,2], iris_features[100:150,3], color="b")
ax.set_xlabel(feature_names[2])
ax.set_ylabel(feature_names[3])
ax.set_title('%s V/s %s' %(feature_names[2], feature_names[3]))
plt.show()
```

```python
# Creating a regression model to predict the petal width, if the petal length is given
# Using linear regression
from sklearn import linear_model
from sklearn.metrics import mean_squared_error


# Create linear regression object
regr = linear_model.LinearRegression()
# Train the model using the training sets
regr.fit(petal_length_train, petal_width_train)

petal_width_pred = regr.predict(petal_length_test)

# test the accuracy
acc_lr = regr.score(petal_length_test, petal_width_test)
print ('prediction accuracy using LR = %0.3f' %(acc_lr))

# The coefficients
print("Coefficients: \n", regr.coef_)
# The mean squared error
print("Mean squared error: %.2f" % mean_squared_error(petal_width_pred, petal_width_test))
```

```python
# Creating a regression model to predict the petal width, if the petal length is given
# using Neural Network - Multi-layer perceptron

from sklearn import neural_network
from sklearn.metrics import mean_squared_error


# Create linear regression object
regr = neural_network.MLPRegressor()
# Train the model using the training sets
regr.fit(petal_length_train, petal_width_train)

petal_width_pred = regr.predict(petal_length_test)

# test the accuracy
acc_mlp = regr.score(petal_length_test, petal_width_test)
print ('prediction accuracy using ANN = %0.3f' %(acc_mlp))

# The mean squared error
print("Mean squared error: %.2f" % mean_squared_error(petal_width_pred, petal_width_test))
```
