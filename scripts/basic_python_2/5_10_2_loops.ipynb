# Loops

Often we may have to repeat some operations many times one after the other. This could be for a fixed number of times or until a condition is statisfied.


## For Loop - Looping For a fixed number of times

**Example 1:**  
You want to print "The best module in this MSc is the Digitisation module!!!" three times.

```
print ("The best module in this MSc is the Digitisation module!!!")
print ("The best module in this MSc is the Digitisation module!!!")
print ("The best module in this MSc is the Digitisation module!!!")
```

In Python, this functionality is usually provided by **For** loop.

For loop usually uses the following syntax to loop for a fixed number of iterations.  

```
for i in range(start_value, end_value, step_size):
    block of code to execute
```

In this, the value of the variable `i` will be initialised with `start_value` and the first iteration of the block of code will be executed.  
In the next iteration, the value of `i` is incremented by the `step_size` and the block of code will be executed again.  
This will continue until the value of `i` remains less than the end_value.

You could any variable name instead of `i`.

Now let's print "The best module in this MSc is the Digitisation module!!!" three times using For loop.

````python
# For loop - example 1-1
for i in range(0, 3, 1):
    print ("i = ", i)
    print ("The best module in this MSc is the Digitisation module!!!")

# The default value for start is 0; default for step_size is 1
# So the above loop can be simplified as
for i in range(3):
    print ("i = ", i)
    print ("The best module in this MSc is the Digitisation module!!!")
````

In some cases, the **number of iterations in the loop may depend on the size of a sequence or group of numbers**, such as a list, or a set.

````python
# For loop - example 1-2
other_msc_modules = ["agri-rootics", "research methods", "product development"]

for i in range(len(other_msc_modules)):
    print ("i = ", i)
    print ("Digitisation is better than ", other_msc_modules[i])

# Here the number of iterations are driven by the length of other_msc_modules
# Try modifying the list with more modules and run it
````

## For Loop - Loop over items of an iterable object

In some special cases, the operation are limited to the items in an iterable object, such as a list of tuple.

When iterating over an iterable object, the syntax of For loop is as follows:  

```
for item in iterable_object:
    do something with item
```

You should not modify the iterable object used for the looping inside the loop. Otherwise, data inconsistency can occur.

Let's consider another way to implement the above example loop.

````python
# For loop - example 2-1
other_msc_modules = ["agri-rootics", "research methods", "product development"]

for module_name in other_msc_modules:
    print ("module = ", module_name)
    print ("Digitisation is better than ", module_name)

# Here the number of iterations are driven by the size of other_msc_modules
# Try modifying the list with more modules and run it
````

Let's consider another example, where we have to find the product of all numbers in a list.


````python
# For loop - example 2-2
a = [3, 4, 2, 8, 7]

a_product = 1
for item in a:
    print ("item: ", item)
    a_product = a_product * item
    print ("a_product: ", a_product)


# a_list is initialised with: [3, 4, 2, 8, 7]
# The for loop was iterated over each item in that list
# i.e., we will take one item from the list from the left, for each iteration
# In iteration 1, item will have a value of 3
# In iteration 2, item will have a value of 4
# In iteration 3, item will have a value of 2
# In iteration 4, item will have a value of 8
# In iteration 5, item will have a value of 7

# Try modifying the list and rerun the cell.
````

## While loop - Looping until a condition remains satisfied

Let's think that we have been given the same list of numbers and were asked to find the smallest group of numbers from the left of this list, which when multiplied will be greater than a threshold value.

Here is one way to do that.  

````python
# An example where multiplication operation is repeated until
# a condition is satisfied

a = [3, 4, 2, 8, 7]

a_threshold = 20

a_product = a[0] * a[1]
if (a_product >= a_threshold):
    print ('The smallest group of numbers is: ', a[0:2])
else:
    a_product = a_product * a[2]
    if (a_product >= a_threshold):
        print ('The smallest group of numbers is: ', a[0:3])
    else:
        a_product = a_product * a[3]
        if (a_product >= a_threshold):
            print ('The smallest group of numbers is: ', a[0:4])
        else:
            a_product = a_product * a[4]
            if (a_product >= a_threshold):
                print ('The smallest group of numbers is: ', a[0:5])
            else:
                print ('The product of the numbers is less than ', a_threshold)

# This one used a lot of condition checking using IF-ELSE.
# We can do this using While loop.
````

The syntax of while loop is as follows:

```
while condition_check:
    # this batch of code will be executed when condition_check is True
    batch of code

    # this batch of code will continue to be executed as many times as possible
    # until condition_check becomes False
```

While loop is a tricky one, as it can result in an infinite loop, if the condition never changes.  
Alternatively, you can use a condition checker inside the loop and use the `break` command to break out of the loop.

You can use `break` in `for` loops as well.

````python
# Here is the above program using WHILE loop
a = [3, 4, 2, 8, 7]
a_threshold = 20

a_product = a[0] * a[1]

i = 2
solution_found = True

while a_product < a_threshold:
    # multiply the next number in the list
    a_product *= a[i]
    # move the index to the next number in the list
    i += 1

    # However, it is possible that the condition is not satisfied even after
    # multiplying all the numbers in the list
    # We have to check for that condition
    if i >= len(a):
        solution_found = False
        break # break out of the while loop if the length of the list is exceeded

if solution_found:
    print ('The smallest group of numbers is: ', a[:i])
else:
    print ('The product of all numbers in the list is less than ', a_threshold)
````
